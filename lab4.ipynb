```python
df['age'] = df['age'].str.replace('31...40','31-40')
df
```

```python
x=df.iloc[:,0:4]
y=df['buys_computer']


x
```

```python
from sklearn.preprocessing import OrdinalEncoder
import pandas as pd

# Reset the datasets to raw split to avoid partial encoding state
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=1
)

# Define distinct encoders for each feature
oe_age = OrdinalEncoder(categories=[['<=30', '31-40', '>40']])
oe_income = OrdinalEncoder(categories=[['low', 'medium', 'high']])
oe_student = OrdinalEncoder(categories=[['no', 'yes']])
oe_credit = OrdinalEncoder(categories=[['fair', 'excellent']])

# Encode both train and test splits to keep them consistent
x_train_encoded = pd.DataFrame(index=x_train.index)
x_train_encoded['age'] = oe_age.fit_transform(x_train[['age']])
x_train_encoded['income'] = oe_income.fit_transform(x_train[['income']])
x_train_encoded['student'] = oe_student.fit_transform(x_train[['student']])
x_train_encoded['credit_rating'] = oe_credit.fit_transform(x_train[['credit_rating']])

x_test_encoded = pd.DataFrame(index=x_test.index)
x_test_encoded['age'] = oe_age.transform(x_test[['age']])
x_test_encoded['income'] = oe_income.transform(x_test[['income']])
x_test_encoded['student'] = oe_student.transform(x_test[['student']])
x_test_encoded['credit_rating'] = oe_credit.transform(x_test[['credit_rating']])

x_train = x_train_encoded
x_test = x_test_encoded

x_train.head()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

y_train = le.fit_transform(y_train)
y_test = le.transform(y_test)
```

```python
from sklearn.naive_bayes import CategoricalNB
```

```python
model = CategoricalNB()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score


acc = accuracy_score(y_test, y_pred)
print(f"Accuracy:{acc:3%}")
```
