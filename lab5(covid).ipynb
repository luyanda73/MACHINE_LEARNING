```python
import pandas as pd
import numpy as np

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder, LabelEncoder
from sklearn.compose import ColumnTransformer

# Load the dataset
df = pd.read_csv('/content/01 covid_toy.csv')

# Display the dataset
print("COVID Dataset:")
display(df)

# Display column names
print("\nColumn Names:")
print(df.columns.tolist())

# Display dataset information
print("\nDataset Information:")
df.info()
```

```python
print("Missing values in each column:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nGender categories:")
print(df["gender"].unique())

print("\nCough categories:")
print(df["cough"].unique())

print("\nCity categories:")
print(df["city"].unique())

print("\nTarget categories:")
print(df["has_covid"].unique())
```

```python
# Input features
X = df.drop("has_covid", axis=1)

# Target column
y = df["has_covid"]

print("Input Features (X):")
display(X)

print("\nTarget Column (y):")
display(y.to_frame())
```

```python
transformer = [
    ('trans1', SimpleImputer(), ['fever']),

    ('trans2',
     OrdinalEncoder(categories=[['Mild', 'Strong']]),
     ['cough']),

    ('trans3',
     OneHotEncoder(sparse_output=False, drop='first'),
     ['gender', 'city'])
]

print("Transformers created successfully.")
```

```python
ct = ColumnTransformer(
    transformers=transformer,
    remainder='passthrough'
)

X_transformed = ct.fit_transform(X)

print("Transformed Input Data:")
display(pd.DataFrame(
    X_transformed,
    columns=ct.get_feature_names_out()
))
```

```python
le = LabelEncoder()

y_encoded = le.fit_transform(y)

print("Original Target Values:")
print(y.tolist())

print("\nEncoded Target Values:")
print(y_encoded)

print("\nLabel Mapping:")
for label, encoded_value in zip(le.classes_, range(len(le.classes_))):
    print(label, "=", encoded_value)
```

```python
# Convert transformed features into a DataFrame
X_final = pd.DataFrame(
    X_transformed,
    columns=ct.get_feature_names_out()
)

# Add the encoded target column
X_final["has_covid"] = y_encoded

print("Final Processed COVID Dataset:")
display(X_final)

print("\nFinal Dataset Shape:")
print(X_final.shape)

print("\nCheck Missing Values:")
display(X_final.isnull().sum().to_frame("Missing Values"))
```
