```python
import pandas as pd

# Load the dataset
file_path = '/00 Social_Network_Ads.csv'
try:
    df = pd.read_csv(file_path)
    print("Dataset successfully loaded!")
    print(f"DataFrame Shape: {df.shape}")
    display(df.head())
except Exception as e:
    print(f"Error loading dataset: {e}")
```

```python
from sklearn.model_selection import train_test_split

# Define features (X) and target variable (y)
X = df[['Age', 'EstimatedSalary']]
y = df['Purchased']

# Perform the train-test split with a test size of 20%
x_train, x_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=1
)

print(f"x_train shape: {x_train.shape}")
print(f"x_test shape: {x_test.shape}")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

# --- 1. Unscaled Model ---
model_unscaled = LogisticRegression(random_state=1)
model_unscaled.fit(x_train, y_train)
y_pred_unscaled = model_unscaled.predict(x_test)
accuracy_unscaled = accuracy_score(y_test, y_pred_unscaled)

# --- 2. Scaled Model ---
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)

model_scaled = LogisticRegression(random_state=1)
model_scaled.fit(x_train_scaled, y_train)
y_pred_scaled = model_scaled.predict(x_test_scaled)
accuracy_scaled = accuracy_score(y_test, y_pred_scaled)

# --- Display Results ---
print(f"Accuracy on unscaled dataframe = {accuracy_unscaled:.4f}")
print(f"Accuracy on scaled dataframe = {accuracy_scaled:.4f}")
```
