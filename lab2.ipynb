```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import SelectKBest, f_regression
from sklearn.decomposition import PCA

```

```python
df=pd.read_csv("auto-mpg.csv")

display(df.head())

```

```python
numeric_columns=df.select_dtypes(include=np.number).columns

print("Numeric Colums:")
print(numeric_columns)


```

```python
categorical_colums=df.select_dtypes(include="object").columns

print("Categorical Colums:")
print(categorical_colums)



```

```python
df["origin"]=df["origin"].astype("object")

```

```python
print(df.isnull().sum())

```

```python
df=df.replace("?",np.nan)
```

```python
print(df.isnull().sum())
```

```python
df['horsepower']=pd.to_numeric(df["horsepower"])
```

```python
df["horsepower"]=df["horsepower"].fillna(df["horsepower"].mean())
```

```python
print(df.isnull().sum())
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler().set_output(transform="pandas")

numeric_cols = df.select_dtypes(include=["number"]).columns

scaled_data = scaler.fit_transform(df[numeric_cols])

scaled_data.head()
```

```python
from sklearn.preprocessing import OneHotEncoder

# Dynamically select object columns to encode (e.g., 'origin', 'car name')
categorical_cols = df.select_dtypes(include=["object"]).columns

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
).set_output(transform="pandas")

encoded_data = encoder.fit_transform(df[categorical_cols])

display(encoded_data.head())
```

```python
import numpy as np
import pandas as pd
from sklearn.feature_selection import SelectKBest, f_regression


y = df["mpg"]

numeric_features = scaled_data.drop(columns=["mpg"], errors="ignore")
X = pd.concat([numeric_features, encoded_data], axis=1)

X_numeric = X.select_dtypes(include=np.number)

selector = SelectKBest(
    score_func=f_regression,
    k=3
).set_output(transform="pandas")

X_selected = selector.fit_transform(X_numeric, y)

print(X_selected.head())
```
